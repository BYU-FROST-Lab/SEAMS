```python
# Load the dictionary of determinants of marginal covariances for each key

with open(Path("../results/nc_tests/quad-easy_first_loop_cov_full6D/firstLoop_ss_cov_full6D.pkl"), "rb") as f:
    sscp_dets = pickle.load(f)
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.close('Same Session Covariance Change Plot')
fig, ax = plt.subplots(1, 1, num='Same Session Covariance Change Plot', figsize=(8, 6))
# ax.set_yscale('log')
# Determine which keys will actually be plotted
items = list(sscp_dets.items())[::5]

# Sample a continuous colormap
colors = plt.colormaps['turbo'](np.linspace(0, 1, len(items)))
# Other nice choices: 'viridis', 'plasma', 'rainbow', 'gist_rainbow'

for (k, dets), color in zip(items, colors):
    # ax[0].plot(dets, color=color, label=f"Key {gtsam.Symbol(k).string()}")
    ax.plot(0.5 * np.log(dets), color=color, label=f"Key {gtsam.Symbol(k).string()}")

ax.legend(loc="right")
ax.set_xlabel("iterations")
ax.set_ylabel("Log Det of 6D Information Matrix")
# ax[1].legend()
plt.tight_layout()
plt.show()
```

```python
# Plot the percentage change

plt.close('Same Session Covariance Relative Percentage Change')
fig, ax = plt.subplots(num='Same Session Covariance Relative Percentage Change',
                       figsize=(10, 6))

items = list(sscp_dets.items())[::5]
colors = plt.colormaps['turbo'](np.linspace(0, 1, len(items)))

for (k, dets), color in zip(items, colors):
    # dets = 0.5 * np.log(np.asarray(dets))
    dets = np.asarray(dets[:100])
    pct_change = 100 * np.diff(dets) / dets[:-1]
    # pct_change = np.diff(dets)

    ax.plot(pct_change, color=color,
            label=f"Key {gtsam.Symbol(k).string()}")

ax.set_ylabel("Relative Change (%)")
ax.set_xlabel("Iteration")
ax.axhline(0, color='k', linestyle='--', linewidth=1)
ax.legend()
plt.show()
```

```python
from numpy.lib.stride_tricks import sliding_window_view

plt.close('Same Session Covariance Rolling Std Dev')
fig, ax = plt.subplots(
    num='Same Session Covariance Rolling Std Dev',
    figsize=(10, 6)
)

items = list(sscp_dets.items())[::5]
colors = plt.colormaps['turbo'](np.linspace(0, 1, len(items)))

window = 5
ax.set_yscale("log")



for (k, dets), color in zip(items, colors):
    dets = np.asarray(dets)

    # Rolling standard deviation over the previous 5 frames
    rolling_std = sliding_window_view(dets, window).std(axis=-1)

    # Shift the x-axis so each value corresponds to the end of its window
    x = np.arange(window - 1, len(dets))

    ax.plot(
        x,
        rolling_std,
        color=color,
        label=f"Key {gtsam.Symbol(k).string()}"
    )

ax.set_ylabel(f"Rolling Std Dev ({window} frames)")
ax.set_xlabel("Iteration")
ax.legend()
plt.show()
```
